## Inverse and determinant

```python
import numpy as np
```

```python
A= np.array([[1,2],[3,4]])
A
```

How to calculate determinant
![How to calculate determinant image](https://media.geeksforgeeks.org/wp-content/uploads/20250425100650715144/Singular-Matrix.webp)

```python
det=np.linalg.det(A)
det
```

```python
B= np.array([[3,1],[6,2]])
B
```

```python
np.linalg.det(B)
```

```python
np.linalg.inv(B)
```
